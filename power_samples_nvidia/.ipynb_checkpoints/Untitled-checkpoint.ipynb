```python
import json
import glob
import os
import matplotlib.pyplot as plt

RESULT_DIR = "results"

json_files = sorted(glob.glob(os.path.join(RESULT_DIR, "*.json")))

if not json_files:
    raise FileNotFoundError(f"No JSON files found in {RESULT_DIR}/")

all_data = []

for path in json_files:
    with open(path, "r") as f:
        data = json.load(f)

    workload = data.get("workload", os.path.splitext(os.path.basename(path))[0])
    results = data.get("results", [])

    freqs = []
    power = []
    exec_ms = []
    measured_freq = []

    for row in results:
        freqs.append(row.get("requested_freq_mhz"))
        measured_freq.append(row.get("avg_measured_freq_mhz"))
        power.append(row.get("avg_power_w"))
        exec_ms.append(
            row.get("avg_execution_ms", row.get("avg_exec_ms"))
        )

    # sort by requested frequency ascending
    zipped = sorted(zip(freqs, measured_freq, power, exec_ms), key=lambda x: x[0])
    freqs, measured_freq, power, exec_ms = map(list, zip(*zipped))

    all_data.append({
        "workload": workload,
        "freqs": freqs,
        "measured_freq": measured_freq,
        "power": power,
        "exec_ms": exec_ms,
    })

# -----------------------------
# Plot 1: Frequency vs Power
# -----------------------------
plt.figure(figsize=(8, 5))
for item in all_data:
    plt.plot(item["freqs"], item["power"], marker="o", label=item["workload"])

plt.xlabel("Requested Frequency (MHz)")
plt.ylabel("Average Power (W)")
plt.title("Frequency vs Power")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# -----------------------------
# Plot 2: Frequency vs Execution Time
# -----------------------------
plt.figure(figsize=(8, 5))
for item in all_data:
    plt.plot(item["freqs"], item["exec_ms"], marker="o", label=item["workload"])

plt.xlabel("Requested Frequency (MHz)")
plt.ylabel("Average Execution Time (ms)")
plt.title("Frequency vs Execution Time")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# -----------------------------
# Plot 3: Requested vs Measured Frequency
# -----------------------------
plt.figure(figsize=(8, 5))
for item in all_data:
    plt.plot(item["freqs"], item["measured_freq"], marker="o", label=item["workload"])

plt.xlabel("Requested Frequency (MHz)")
plt.ylabel("Measured Frequency (MHz)")
plt.title("Requested vs Measured Frequency")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import json
import glob
import os
import numpy as np
import matplotlib.pyplot as plt

# --------- CONFIG ----------
RESULT_DIR = "results"
TARGET_FREQ = 1200  # The frequency you want execution times for
# ---------------------------

json_files = sorted(glob.glob(os.path.join(RESULT_DIR, "*.json")))

if not json_files:
    raise FileNotFoundError(f"No JSON files found in {RESULT_DIR}/")

all_data = []

for path in json_files:
    with open(path, "r") as f:
        data = json.load(f)

    workload = data.get("workload", os.path.splitext(os.path.basename(path))[0])
    results = data.get("results", [])

    freqs = []
    power = []
    exec_ms = []
    measured_freq = []

    for row in results:
        freqs.append(row.get("requested_freq_mhz"))
        measured_freq.append(row.get("avg_measured_freq_mhz"))
        power.append(row.get("avg_power_w"))
        exec_ms.append(
            row.get("avg_execution_ms", row.get("avg_exec_ms"))
        )

    # Sort data by requested frequency ascending
    zipped = sorted(zip(freqs, measured_freq, power, exec_ms), key=lambda x: x[0])
    freqs, measured_freq, power, exec_ms = map(list, zip(*zipped))

    all_data.append({
        "workload": workload,
        "freqs": freqs,
        "measured_freq": measured_freq,
        "power": power,
        "exec_ms": exec_ms,
    })

# ---------------------------------------------------------
# NEW SECTION: Get Execution Time for Specific Frequency
# ---------------------------------------------------------
print(f"\n{'='*60}")
print(f"EXECUTION TIMES AT {TARGET_FREQ} MHz")
print(f"{'='*60}")
print(f"{'Workload':<25} | {'Matched Freq':<15} | {'Exec Time (ms)':<15}")
print("-" * 60)

for item in all_data:
    freqs_arr = np.array(item["freqs"])
    
    # Find the index of the frequency closest to TARGET_FREQ
    idx = (np.abs(freqs_arr - TARGET_FREQ)).argmin()
    
    matched_f = item["freqs"][idx]
    execution_time = item["exec_ms"][idx]
    
    # Flag if the match isn't exact
    exact_str = "" if matched_f == TARGET_FREQ else " (approx)"
    
    print(f"{item['workload']:<25} | {matched_f:<8} MHz{exact_str:<5} | {execution_time:<15.4f}")

print(f"{'='*60}\n")
```
