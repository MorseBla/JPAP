```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# ================== CONFIG ==================
# Your runs live under: logs/mm_mm/setpoint_{SP}{TAG}/
# Example: logs/mm_mm/setpoint_0.80open/FC_GPU_t1_rtr
root = Path("logs/mm_mm")
setpoints = [0.75, 0.80, 0.85, 0.90]
open_tag = "open"

t1_name = "FC_GPU_t1_rtr.txt"
t2_name = "FC_GPU_t2_rtr.txt"

# If you want to save a figure:
save_path = None  # e.g. Path("all_setpoints.png")
# ===========================================

def read_floats(path: Path) -> np.ndarray:
    vals = []
    if not path.exists():
        raise FileNotFoundError(f"Missing file: {path}")
    with path.open("r") as f:
        for line in f:
            s = line.strip()
            if not s:
                continue
            try:
                vals.append(float(s))
            except ValueError:
                continue
    if not vals:
        raise ValueError(f"No numeric samples found in {path}")
    return np.asarray(vals, dtype=np.float64)

def fit_to_n(x: np.ndarray, n: int) -> np.ndarray:
    if len(x) < n:
        return np.pad(x, (0, n - len(x)), mode="edge")
    return x[:n]

# Plot all setpoints on one figure
plt.figure(figsize=(12, 5))

for sp in setpoints:
    # Folder name pattern: setpoint_{SP}{TAG}  (e.g., setpoint_0.80open)
    run_dir = root / f"setpoint_{sp:.2f}_{sp:.2f}"

    t1_path = run_dir / t1_name
    t2_path = run_dir / t2_name

    # Load signals
    t1 = read_floats(t1_path)
    t2 = read_floats(t2_path)

    n = int(max(len(t1), len(t2)))
    x = np.arange(n)

    t1f = fit_to_n(t1, n)
    t2f = fit_to_n(t2, n)
    sp_line = np.full(n, sp, dtype=np.float64)

    # Plot (label includes setpoint so curves are distinguishable)
    plt.plot(x, t1f, label=f"{t1_name} (sp={sp:.2f})")
    plt.plot(x, t2f, label=f"{t2_name} (sp={sp:.2f})")
    plt.plot(x, sp_line, linestyle="--", label=f"setpoint {sp:.2f}")

plt.xlabel("sample")
plt.ylabel("value")
plt.title("Experiment logs vs setpoint (multiple setpoints)")
plt.grid(True)
plt.legend(ncols=2, fontsize=9)
plt.tight_layout()

if save_path is not None:
    plt.savefig(save_path, dpi=200)
    print(f"Saved: {save_path}")

plt.show()
```
